**3. Idade, Desempregado ou aposentado e Anos_Trabalhados**

As colunas DAYS_BIRTH e DAYS_EMPLOYED estão  em dias e com valores negativos. Criamos nova coluna "Idade" com o valor em anos, considerando o ano bissexto e alteramos o valor de negativo para positivo.

O valor positivo 365243 do DAYS_EMPLOYED é uma nomalia na base de dados que costuma representar pessoas que estão atualmente desempregadas ou aposentadas (já que os dias de trabalho ativos são registrados como números negativos). Por isso, criamos nova coluna chamada Desempregados ou aposentados, onde, se o valor for exatamente 365243, a nova coluna receberá "True".

Criamos também mais uma coluna "Anos_Trabalhados" que mostra quantos anos cada pessoa tem de serviço, tratando o valor positivo 365243 igual a 0 anos.

```python
df_excel['Idade'] = (df_excel['DAYS_BIRTH'] * -1) / 365.25
```

```python
df_excel['Desempregado ou aposentado'] = df_excel['DAYS_EMPLOYED'] == 365243
```

```python
df_excel.info()
```

```python
df_excel['Anos_Trabalhados'] = df_excel['DAYS_EMPLOYED'].apply(lambda x: 0.0 if x == 365243 else (x * -1) / 365.25)
```

**9. "Mau pagador" e "bom pagador"**

O código abaixo cria um DataFrame resumido por `ID` contendo a classificação de risco (Target) para cada cliente, onde definimos como **Mau Pagador (1)** os clientes com atraso superior a 60 dias (Status 2, 3, 4 ou 5) no histórico, e o restante como **Bom Pagador (0)**.

Pela classificação, dos 45985 clientes (ID únicos), 45318 são bons pagadores e 667 maus pagadores.

```python
df_excel2['MAU_PAGADOR'] = df_excel2['STATUS'].isin(['2', '3', '4', '5']).astype(int)
df_target = df_excel2.groupby('ID')['MAU_PAGADOR'].max().reset_index()
```

```python
print("Proporção de Clientes Bons (0) vs Maus (1) consolidada:")
display(df_target['MAU_PAGADOR'].value_counts())
display(df_target['MAU_PAGADOR'].value_counts(normalize=True) * 100)
```
